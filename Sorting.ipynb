# Sorting

Sorting a list makes binary search, finding median, checking dublicates, building frequency tables, etc easier.

```python
import time
import random
import sys
sys.setrecursionlimit(2**31-1)
```

```python
def timer(f, *args):
  start = time.perf_counter()
  print(f(*args))
  end = time.perf_counter()
  print(end-start)
```

```python
def naive_search(li,s):
  for e in li:
    if e==s:
      return True
  return False
```

```python
def binary_search(li,s):
  if len(li) == 0:
    return False

  i = len(li)//2
  if li[i] == s:
    return True
  elif s<li[i]:
    return binary_search(li[:i],s)
  else:
    return binary_search(li[i+1:],s)
```

```python
a = [random.randint(1,100) for i in range(0,100)]

sa = sorted(a)

timer(naive_search, a,1)
timer(binary_search,sa,1)
```

```python
from IPython.display import clear_output
def list_tracker(li):
  clear_output(wait=True)
  for i in li:
    print("*"*i)
  time.sleep(0.1)
```

## Selection sort

```python
def find_min(li):
  prev_index,prev = 0,li[0]
  for i in range(1,len(li)):
    if li[i] < prev:
      prev_index, prev = i,li[i]
  return prev_index, prev


def selection_sort(li):
  sli = []
  li_len = len(li)

  for i in range(0,li_len):
    min_index, min = find_min(li)
    sli.append(min)
    li.pop(min_index)


  return sli
```

```python
# algorithm that does to keep a new sorted list
def selection_sort(li):
  k=0
  list_tracker(li)
  for i in range(len(li)):
    # we assume that li[:i] (list does not include i'th element) is sorted
    # moreover all elements of li[:i] are less than all elements of li[i:]
    # we then set i'th element to be the least element in the list [i:]

    for j in range(i+1,len(li)):
      # if j'th element is less than i'th element, then we replace i'th
      # position by j'th position
      k+=1
      if li[i] > li[j]:
        li[i] , li[j] = li[j], li[i]
        list_tracker(li)

    # on the completion of the loop, we replaced i'th element with the
    # least element in the list [i:]. Note that li[:i+1] is sorted,
    # as li[:i] was sorted and l[i] is greater than all elments in l[:i].
    # Moreover every element in li[i+1:] is greater than every element
    # in li[:i+1], because every element in li[:i] is less than every
    # element in li[i:] and so in li[i+1:], moreover l[i] is the smallest
    # element in li[i:] hence every element in li[i+1:] is greater than
    # li[i]

  return li,k
```

T(n) &= n + (n-1) + \cdots + (n-(n-1)) \\ &= n^2 - ((n-1)n)/2 \\ &= n(n+1)/2
\end{align}$$
+ $T(n)$ is $O(n)$.
+ Every case is the worst case, i.e every case iterates $n$ times.

## Insertion sort

```python
def insertion_sort(li):
  sli = [li[0]]
  k=0
  for i in range(1,len(li)):
    for j in range(i-1, -1, -1):
      k+=1

      if li[i] >= sli[j]:
        sli.insert(j+1,li[i])
        break
      elif j==0:
        sli.insert(0,li[i])

  return sli,k
```

```python
# without creating a new list
def insertion_sort(li):
  list_tracker(li)
  k=0
  for i in range(1,len(li)):
    for j in range(i-1, -1, -1):

      if li[j+1] >= li[j]:
        break

      k+=1

      li[j], li[j+1] = li[j+1], li[j]
      list_tracker(li)

  return li,k
```

```python
# using while loop

def insertion_sort(li):
  k=0

  for i in range(1,len(li)):
    j = i
    while li[j] < li[j-1] and j > 0:
      k+=1
      li[j],li[j-1] = li[j-1],li[j]
      j-=1

  return li,k
```

```python
# more intuitive insertion sort

def insertion_sort(li):
  if len(li) == 1:
    return li
  return insert(insertion_sort(li[:-1]),li[-1])

def insert(li, e):
  if len(li)==0:
    return [e]
  elif li[-1] <= e:
    return li + [e]
  else:
    return insert(li[:-1],e) + li[-1:]
```

$$T(n) = 1 + 2 + \cdots + n-1 =  n(n-1)/2$$ by while loop algorithm.

By recusive algorithm,
$$T_{\text{sort}}(1) = 1, \\ T_{\text{sort}}(n) = T_{\text{sort}}(n-1) + T_{\text{insert}}(n-1),$$
and $$T_{\text{insert}}(0) = T_{\text{insert}}(1) = 1, \\ T_{\text{insert}}(n) = 1 + T_{\text{insert}}(n-1) = n.$$
Hence, $$\begin{align}T_{\text{sort}}(n) &= T_{\text{sort}}(n-2) + T_{\text{insert}}(n-2) + n-1 \\ &= 1 + 2 + \cdots + n-1 \\ &= n(n-1)/2.\end{align}$$


+ $T(n)$ is $O(n^2)$ for worst case
+ Unlike selection sort, every case is not the worst case, some are better.

## Merge sort

```python
def merge_sort(li):
  if len(li) == 1:
    return li

  m = len(li)//2

  return merge(merge_sort(li[:m]),merge_sort(li[m:]))

```

```python
def merge(li1,li2):
  i,j,k = 0,0,0
  M = []
  while k < len(li1) + len(li2):

    if i == len(li1):
      M.extend(li2[j:])
      break
    elif j == len(li2):
      M.extend(li1[i:])
      break
    elif li1[i] < li2[j]:
      M.append(li1[i])
      i+=1
    else:
      M.append(li2[j])
      j+=1

    k+=1
  return M
```

```python
# Variation of merge (removes dublicate)
def merge(li1,li2):
  i,j,k = 0,0,0
  M = []
  while k < len(li1) + len(li2):

    if i == len(li1):
      M.extend(li2[j:])
      break
    elif j == len(li2):
      M.extend(li1[i:])
      break
    elif li1[i] == li2[j]:
      M.append(li1[i])
      i+=1
      j+=1
    elif li1[i] < li2[j]:
      M.append(li1[i])
      i+=1
    else:
      M.append(li2[j])
      j+=1

    k+=1
  return M
```

```python
merge_sort([2,3,4,2,3,5,4])
```

$$\begin{align}T_{\text{sort}}(n) &\approx 2T_{\text{sort}}(n/2)+T_{\text{merge}}(n/2,n/2) = 2T_{\text{sort}}(n/2)+n \\ &\approx 2(2T_{\text{merge}}(n/4) + n/2) + n = 2^2 T(n/4) + n + n \\ &\approx 2^{\log_{2}n} T(n/\log_2(n)) + \log_2(n) n = n T_{\text{merge}}(0) + n\log n \\ &= n + n\log n \end{align} \\ $$

+ $T(n)$ is $O(n\log_2(n))$ in the worst case
+ Takes too much storage
+ recursive calls are expensive

## Quick sort

```python
def quick_sort(li):
  if len(li) == 0:
    return li

  p = li[0]
  prep = [p]
  l = []
  r = []

  for e in li[1:]:
    if e < p:
      l.append(e)
    elif e > p:
      r.append(e)
    else:
      prep.append(p)

  return quick_sort(l) + prep + quick_sort(r)

# but we created so many lists in the storage
```

Using partitioning:
[pivot|lower|upper|unclassified]

```python
def quick_sort(li, a=None, b=None): # sort li[l:r]

  if a == None and b == None: a,b = 0, len(li)

  if b-a <= 1:
    return li

  # pivot, lower next, unclassified
  p, ln, uc = li[a], a+1, a+1

  for i in range(a+1, b):
    if li[i] > p:
      uc+=1
    else:
      li[ln], li[i] = li[i], li[ln]
      ln+=1
      uc+=1

  # move pivot between lower and upper

  li[a], li[ln-1] = li[ln-1], li[a]

  quick_sort(li,a,ln-1)
  quick_sort(li,ln, b)
  return li
```

```python
li = [random.randrange(10**5+10**2) for i in range(10**5)]
cli = li
asd = [i for i in range(10**4)]
casd = asd
dsd = [i for i in range(10**4-1,-1,-1)]
cdsd = dsd
print("merge sort of already sorted list")
timer(merge_sort,asd)
print("quick sort on already sorted list")
timer(quick_sort,casd),
print("merge sort on descending list")
timer(merge_sort,dsd)
print("quick sort on descending list")
timer(quick_sort,cdsd)
print("merge sort on random list")
timer(merge_sort,li)
print("quick sort on random list")
timer(quick_sort,cli)
# notice the time taken by quick sort on random list is half that for merge sort
```

Notice the algorithm works like merge sort (having $O(\log_2(n))$ complexity) when the pivot is actually the median (thus dividing the list in literal half baring in mind repitions). But in the algorithm, we always choose the first element of the list as the pivot, if a bad actor gives use the list that is already sorted, then the complexity that we get is $O(n^2)$. Even though it that is the case, for a random list (or on choosing random pivot), we get the expectation to be $O(\log_2(n))$, while performing faster than merge sort and it does not really has merge process (simple connecting works). Note that if we try to predict in advance,the median, we are required to employ sorting of the list without knowing it, thus defeating the whole of purpose of sorting.
